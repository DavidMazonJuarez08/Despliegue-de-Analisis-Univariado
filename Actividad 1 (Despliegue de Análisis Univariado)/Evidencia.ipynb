```python
# mostrar los meses encontrados
for fila in filas_meses:
    print(fila, FUNNEL4.iloc[fila, 0])
```

```python
#Encontrar los encabezados 
for posicion, fila_mes in enumerate(filas_meses):

    # Nombre del mes
    mes = str(FUNNEL4.iloc[fila_mes, 0]).replace("Funnel ", "").strip()

    # Fin del bloque del mes
    if posicion + 1 < len(filas_meses):
        fin_mes = filas_meses[posicion + 1]
    else:
        fin_mes = len(FUNNEL4)

    # Buscar fila donde aparecen los encabezados "Item"
    fila_encabezado = None

    for fila in range(fila_mes + 1, min(fila_mes + 5, fin_mes)):

        cantidad_item = sum(
            str(valor).strip().lower() == "item"
            for valor in FUNNEL4.iloc[fila].dropna()
        )

        if cantidad_item >= 2:
            fila_encabezado = fila
            break

    print(mes, "-> fila encabezado:", fila_encabezado)
```

```python
registros = []

for posicion, fila_mes in enumerate(filas_meses):

    # Nombre del mes
    mes = str(FUNNEL4.iloc[fila_mes, 0]).replace("Funnel ", "").strip()

    # Determinar hasta dónde llega el bloque del mes
    if posicion + 1 < len(filas_meses):
        fin_mes = filas_meses[posicion + 1]
    else:
        fin_mes = len(FUNNEL4)

    # Buscar fila de encabezados
    fila_encabezado = None

    for fila in range(fila_mes + 1, min(fila_mes + 5, fin_mes)):

        cantidad_item = sum(
            str(valor).strip().lower() == "item"
            for valor in FUNNEL4.iloc[fila].dropna()
        )

        if cantidad_item >= 2:
            fila_encabezado = fila
            break

    fila_canales = fila_encabezado - 1

    # Buscar cada bloque que comienza con "Item"
    for columna in range(FUNNEL4.shape[1]):

        encabezado = str(
            FUNNEL4.iloc[fila_encabezado, columna]
        ).strip().lower()

        if encabezado == "item":

            canal = FUNNEL4.iloc[fila_canales, columna]

            if pd.isna(canal):
                continue

            canal = str(canal).strip()

            # Recorrer las filas del funnel
            for fila in range(fila_encabezado + 1, fin_mes):

                item = FUNNEL4.iloc[fila, columna]

                if pd.isna(item):
                    continue

                item = str(item).strip()

                if item == "":
                    continue

                # Obtener las variables del bloque
                obj = FUNNEL4.iloc[fila, columna + 1] if columna + 1 < FUNNEL4.shape[1] else np.nan
                real = FUNNEL4.iloc[fila, columna + 2] if columna + 2 < FUNNEL4.shape[1] else np.nan
                acumulado = FUNNEL4.iloc[fila, columna + 3] if columna + 3 < FUNNEL4.shape[1] else np.nan
                obj_al_dia = FUNNEL4.iloc[fila, columna + 4] if columna + 4 < FUNNEL4.shape[1] else np.nan
                cumplimiento = FUNNEL4.iloc[fila, columna + 5] if columna + 5 < FUNNEL4.shape[1] else np.nan

                registros.append([
                    mes,
                    canal,
                    item,
                    obj,
                    real,
                    acumulado,
                    obj_al_dia,
                    cumplimiento
                ])
```

```python
df_funnel = pd.DataFrame(registros, columns=["Mes", "Canal", "Item", "Obj", "Real", "Acumulado", "Obj_al_dia", "Cumplimiento"])
```

```python
meses_orden = ['Enero','Febrero','Marzo','Abril','Mayo','Junio','Julio ','Agosto',
               'Septiembre','Octubre','Noviembre','Diciembre']

def extraer_bloque(nombre_bloque, filas_dict):
    #Extrae la fila "Real" de cada bloque del dashboard (12 meses, columnas 1 a 12)
    datos = {}
    for etiqueta, fila in filas_dict.items():
        datos[f"{nombre_bloque}_{etiqueta}"] = TOPS6.iloc[fila, 1:13].tolist()
    return datos

bloques = {}
bloques.update(extraer_bloque('Afluencia',        {'RealPisos': 3}))
bloques.update(extraer_bloque('VentasPlaza',       {'RealVentas': 15}))
bloques.update(extraer_bloque('RedesSociales',     {'RealVisualizaciones': 26}))
bloques.update(extraer_bloque('UtilidadNeta',      {'Real': 36}))
bloques.update(extraer_bloque('RotacionPersonal',  {'Rotacion': 45}))
bloques.update(extraer_bloque('RecuperacionMarca', {'Real': 55}))

tops = pd.DataFrame(bloques, index=[m.strip() for m in meses_orden])
tops
```

```python
def limpiar_valor(v):
    #Convertimos strings con "%" o vacios a tipo numerico (float)
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = v.strip().replace('%', '')
        if v == '':
            return np.nan
        return float(v)
    return float(v)

for c in tops.columns:
    tops[c] = tops[c].apply(limpiar_valor)

tops.info()
```

```python
# Detectar valores atípicos mediante el método IQR

# Seleccionar únicamente las variables numéricas
numericas = marketing_2025_2026.select_dtypes(include='number')

# Calcular el primer y tercer cuartil
Q1 = numericas.quantile(0.25)
Q3 = numericas.quantile(0.75)

# Calcular el rango intercuartílico
IQR = Q3 - Q1

# Definir los límites inferior y superior
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Contar cuántos posibles outliers hay en cada variable
outliers = (numericas < limite_inferior) | (numericas > limite_superior)

outliers.sum()
```

```python
# Mostrar las observaciones identificadas como posibles outliers
marketing_2025_2026[outliers.any(axis=1)]
```

```python
# Mostrar exactamente qué valores fueron detectados como outliers
for columna in outliers.columns:
    if outliers[columna].sum() > 0:
        print(f"\n{columna}:")
        print(
            marketing_2025_2026.loc[
                outliers[columna],
                ['Mes', columna]
            ]
        )
```

Al aplicar el método IQR identifiqué algunos valores atípicos en diferentes variables. 
Sin embargo, decidí conservarlos, ya que al revisar los datos 
#considero que pueden representar variaciones reales entre los meses y no necesariamente errores en la base.

```python
#Metodo IQR: encontramos los limites y contamos outliers por variable
#(son variables binarias 0/1, una escala 0-5 (0="sin evaluar"), por lo que el metodo IQR marca
#como "atipica" a la categoria minoritaria (ej. los pocos "Venta=1"), no un dato erroneo; por
#eso NO se eliminan ni se recortan, son valores validos y necesarios para el analisis de negocio)
for c in numericas_bit:
    Q1 = bit_clean[c].quantile(0.25)
    Q3 = bit_clean[c].quantile(0.75)
    IQR = Q3 - Q1
    li = Q1 - 1.5 * IQR
    ls = Q3 + 1.5 * IQR
    outliers = bit_clean[(bit_clean[c] < li) | (bit_clean[c] > ls)]
    print(f"{c}: limites=({li:.2f}, {ls:.2f}) | outliers detectados={len(outliers)}")
```

```python
#Metodo IQR: detectamos limites y outliers, y aplicamos "capping" (winsorizacion)
#Se usa capping (no eliminacion) porque son solo 12 meses de datos y perder un mes
#rompe el analisis de tendencia
tops_capado = tops_clean.copy()
for c in tops_clean.columns:
    Q1 = tops_clean[c].quantile(0.25)
    Q3 = tops_clean[c].quantile(0.75)
    IQR = Q3 - Q1
    li = Q1 - 1.5 * IQR
    ls = Q3 + 1.5 * IQR
    outliers = tops_clean[(tops_clean[c] < li) | (tops_clean[c] > ls)]
    print(f"{c}: limites=({li:.1f}, {ls:.1f}) | outliers={len(outliers)}")
    tops_capado[c] = tops_capado[c].clip(lower=li, upper=ls)

tops_clean = tops_capado
```

```python
# Calcular el número de observaciones de Leads total
n = marketing_2025_2026['Leads total'].count()
n
```

```python
# Calcular el rango de Leads total
rango = maximo - minimo
rango
```

```python
# Calcular el número de intervalos con la regla de Sturges
k = 1 + 3.322 * np.log10(n)
k
```

```python
# Redondear el número de intervalos
k = round(k)
k
```

```python
# Calcular la amplitud de cada intervalo
amplitud = rango / k
amplitud
```

```python
# Crear los límites de los 5 intervalos
limites = np.linspace(minimo, maximo, k + 1)

limites
```

```python
#Analisis univariado de la variable cualitativa "Asesor": leads atendidos por cada asesor
Tabla_freq_asesor = bit_clean['Asesor'].value_counts().reset_index()
Tabla_freq_asesor.columns = ['Asesor', 'count']
Tabla_freq_asesor
```

```python
#Grafico de barras del volumen de leads por asesor
Tabla_freq_asesor.set_index('Asesor').plot(kind='bar', width=0.8, figsize=(12,4), color="red")
plt.title('Leads atendidos por Asesor')
plt.xlabel('Asesor')
plt.ylabel('Frecuencia (leads)')
plt.tight_layout()
plt.show()
```

```python
#DECISION ESTRATEGICA: no aplicamos la Regla de Sturges a Temperatura_Nivel, PDM, SDC, Venta
#o Inter Gerente porque son variables binarias (0/1) o una escala ordinal fija de 0 a 5;
#ya estan categorizadas y partirlas en mas clases no aporta informacion.
#En cambio, el numero de leads por asesor (Tabla_freq_asesor['count']) si es una variable
#cuantitativa que conviene agrupar en clases para identificar niveles de carga de trabajo.
n = len(Tabla_freq_asesor)
Max = Tabla_freq_asesor['count'].max()
Min = Tabla_freq_asesor['count'].min()
R = Max - Min
ni = 1 + 3.32*np.log10(n)
i = R/ni
print("n:", n, "| R:", R, "| ni:", round(ni), "| i:", round(i, 2))
```

```python
#Creamos los intervalos y las categorias de volumen de leads por asesor
#include_lowest=True asegura que el asesor con el valor minimo tambien quede incluido
#en la primera clase (equivalente a "ampliar los limites" del ejemplo de clase)
intervalos = np.linspace(Min, Max+1, round(ni)+1)
categorias = [f"{intervalos[k]:.0f}-{intervalos[k+1]:.0f}" for k in range(len(intervalos)-1)]
Tabla_freq_asesor['Volumen_Leads'] = pd.cut(x=Tabla_freq_asesor['count'], bins=intervalos,
                                             labels=categorias, include_lowest=True)
Tabla_freq_asesor
```

```python
#Analisis univariado de la nueva variable categorica creada
Tabla_categoria = Tabla_freq_asesor['Volumen_Leads'].value_counts().reset_index()
Tabla_categoria.columns = ['Volumen_Leads', 'count']
Tabla_categoria.set_index('Volumen_Leads').plot(kind='bar', width=0.8, figsize=(8,4), color="red")
plt.title('Asesores agrupados por volumen de leads')
plt.xlabel('Categoria de volumen')
plt.ylabel('Numero de asesores')
plt.tight_layout()
plt.show()
```

```python
#Integramos la categoria de volumen de leads al DataFrame principal (union por Asesor)
bit_clean = bit_clean.merge(Tabla_freq_asesor[['Asesor', 'Volumen_Leads']], on='Asesor', how='left')
bit_clean[['Asesor', 'Volumen_Leads']].head()
```

```python
#Regresion Lineal Simple: Apvs_Cumplieron (absoluto) ~ Total_Apvs
Var_Dep = df_metas['Apvs_Cumplieron']
Vars_Indep = df_metas[['Total_Apvs']]

model_metas_abs = LinearRegression()
model_metas_abs.fit(X=Vars_Indep, y=Var_Dep)

m = model_metas_abs.coef_[0]
b = model_metas_abs.intercept_
coef_Deter = model_metas_abs.score(Vars_Indep, Var_Dep)
coef_Correl = np.sqrt(coef_Deter) if m >= 0 else -np.sqrt(coef_Deter)

print(f"Modelo matematico: Apvs_Cumplieron = {m:.4f} * Total_Apvs + {b:.4f}")
print(f"Coeficiente de Determinacion R2 = {coef_Deter:.4f}")
print(f"Coeficiente de Correlacion r = {coef_Correl:.4f}")
```

```python
#Predecimos y graficamos real vs prediccion (valores absolutos)
df_metas['Predicciones_Abs'] = model_metas_abs.predict(Vars_Indep)

sns.scatterplot(x='Total_Apvs', y='Apvs_Cumplieron', color='blue', data=df_metas)
sns.scatterplot(x='Total_Apvs', y='Predicciones_Abs', color='red', data=df_metas)
sns.lineplot(x='Total_Apvs', y='Predicciones_Abs', color='red', data=df_metas)
plt.title('Apvs que cumplieron: real (azul) vs. predicho (rojo) según Total de Apvs')
plt.show()
```

```python
#Dispersion antes de ajustar el modelo: % de Cumplimiento (tasa) vs Total de Apvs
#Esta es la variable que realmente responde la pregunta de "escalabilidad"
sns.scatterplot(x='Total_Apvs', y='Pct_Cumplimiento', color='blue', data=df_metas)
plt.title('Dispersión: % de Cumplimiento vs Total de Apvs')
plt.show()
```

```python
#Regresion Lineal Simple: % Cumplimiento ~ Total de Apvs (esta es la que responde "escalabilidad")
Var_Dep = df_metas['Pct_Cumplimiento']
Vars_Indep = df_metas[['Total_Apvs']]

model_metas = LinearRegression()
model_metas.fit(X=Vars_Indep, y=Var_Dep)

m = model_metas.coef_[0]
b = model_metas.intercept_
coef_Deter = model_metas.score(Vars_Indep, Var_Dep)
coef_Correl = np.sqrt(coef_Deter) if m >= 0 else -np.sqrt(coef_Deter)

print(f"Modelo matematico: %Cumplimiento = {m:.4f} * Total_Apvs + {b:.4f}")
print(f"Coeficiente de Determinacion R2 = {coef_Deter:.4f}")
print(f"Coeficiente de Correlacion r = {coef_Correl:.4f}")

if coef_Correl <= -0.3:
    conclusion = "El cumplimiento de metas NO es escalable: conforme crece el equipo, el % de asesores que cumplen su meta tiende a bajar (dilucion)."
elif coef_Correl >= 0.3:
    conclusion = "El cumplimiento de metas SI escala bien: conforme crece el equipo, el % de asesores que cumplen su meta se mantiene o mejora."
else:
    conclusion = "El tamaño del equipo no tiene una relacion lineal fuerte con la TASA de cumplimiento: crece el numero absoluto de asesores que cumplen, pero el porcentaje se mantiene relativamente estable (ni mejora ni se diluye claramente)."
print("\nConclusion:", conclusion)
```

```python
#Predecimos y graficamos real vs prediccion (tasa de cumplimiento)
df_metas['Predicciones_Pct'] = model_metas.predict(Vars_Indep)

sns.scatterplot(x='Total_Apvs', y='Pct_Cumplimiento', color='blue', data=df_metas)
sns.scatterplot(x='Total_Apvs', y='Predicciones_Pct', color='red', data=df_metas)
sns.lineplot(x='Total_Apvs', y='Predicciones_Pct', color='red', data=df_metas)
plt.title('% de Cumplimiento: real (azul) vs. predicho (rojo) según Total de Apvs')
plt.show()
```
